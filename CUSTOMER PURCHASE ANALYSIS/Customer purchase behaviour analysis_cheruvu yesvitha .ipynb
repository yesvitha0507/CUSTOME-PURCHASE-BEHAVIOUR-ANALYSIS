### Convert Date column to datetime

```python
df['Date'] = pd.to_datetime(df['Date'])
```

### Calculate Total Amount (Price * Quantity)

```python
df['TotalAmount'] = df['Price'] * df['Quantity']
```

### INSIGHTS AND SUMMARY REPORTS

### Total Revenue

```python
print(df['TotalAmount'].sum())
```

### REVENUE BY GENDER

```python
print(df.groupby('Gender')['TotalAmount'].sum())
```

### REVENUE BY CITY

```python
print(df.groupby('City')['TotalAmount'].sum())
```

### MOST PURCHASED PRODUCTS (Quantity-wise)

```python
print(df.groupby('Product')['Quantity'].sum().sort_values(ascending=False))
```

### AGE GROUP PURCHASE COUNT

```python
bins = [0, 20, 30, 40]
labels = ['0-20', '21-30', '31-40']
df['AgeGroup'] = pd.cut(df['Age'], bins=bins, labels=labels, right=False)
print(df.groupby('AgeGroup',observed=True)['TotalAmount'].sum())
```

### ADVANCED INSIGHTS

### CATEGORY-WISE REVENUE

```python
print(df.groupby('Category')['TotalAmount'].sum())
```

### CUSTOMER-WISE PURCHASE TOTAL 

```python
print(df.groupby('Name')['TotalAmount'].sum())
```

###  PURCHASES BY AGE GROUP (Count) 

```python
print(df.groupby('AgeGroup',observed=True)['Product'].count())
```

### HIGH VALUE PURCHASE % (>10000)

```python
high_value = np.where(df['TotalAmount'] > 10000, 1, 0)
print("Percentage:", (high_value.sum() / len(df)) * 100, "%")
```

### DAILY PURCHASE TREND

```python
print(df.groupby('Date')['TotalAmount'].sum())
```
